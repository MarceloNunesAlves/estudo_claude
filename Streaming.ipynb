```python
from dotenv import load_dotenv
load_dotenv(override=True)

from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system_prompt=None):
    parameters = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "stream": True
    }

    if system_prompt:
        parameters["system"] = system_prompt

    message = client.messages.create(**parameters)
    return message
```

```python
messages = []
input_text = "Como faço um bolo de chocolate?"
print(f"User: {input_text}")
add_user_message(messages, input_text)
stream = chat(messages)
for event in stream:
    print(event)
```

```python
messages = []
input_text = "Quais são os 10 pontos turísticos de São Paulo?"
print(f"User: {input_text}")
add_user_message(messages, input_text)
with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:
    for text in stream.text_stream:
        print(text, end='')

    final_message = stream.get_final_message()
```
