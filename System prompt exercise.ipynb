```python
from dotenv import load_dotenv
load_dotenv(override=True)

from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system_prompt=None):
    parameters = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system_prompt:
        parameters["system"] = system_prompt

    message = client.messages.create(**parameters)
    return message.content[0].text if getattr(message.content[0], 'text', None) else message.content[1].text
```

```python
# Start with an empty message list
messages = []
```

```python
input_text = "função de fibonacci"
print(f"User: {input_text}")
add_user_message(messages, input_text)
answer = chat(messages)
add_assistant_message(messages, answer)
print(answer)
```

```python
system_prompt = "Você é um desenvolvedor de sotfware em Python e deve escrever código sem comentarios. Você deve responder apenas com o código solicitado, sem explicações ou comentários adicionais. Se a pergunta não for sobre código, você deve responder com 'Desculpe, não posso ajudar com isso.'"

messages = []

input_text = "função de fibonacci"
print(f"User: {input_text}")

add_user_message(messages, input_text)
answer = chat(messages, system_prompt=system_prompt)
add_assistant_message(messages, answer)
print(answer)
```
