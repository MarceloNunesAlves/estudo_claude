````python
from dotenv import load_dotenv
load_dotenv(override=True)

from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"
````

````python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system_prompt=None, stop_sequences=None):
    parameters = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }

    if system_prompt:
        parameters["system"] = system_prompt

    if stop_sequences:
        parameters["stop_sequences"] = stop_sequences

    message = client.messages.create(**parameters)
    return message
````

````python
messages = []
input_text = "Generate three different sample AWS cli commands. Each should be very short."
print(f"User: {input_text}")
add_user_message(messages, input_text)
add_assistant_message(messages, "Here is three different commands in a single block without comments:\n```bash")
answer = chat(messages, stop_sequences=["```"])
answer
````
